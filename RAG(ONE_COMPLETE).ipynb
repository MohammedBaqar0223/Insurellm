```python
load_dotenv()
MODEL_1 = "gemini-2.5-flash"
gemini = OpenAI(api_key=os.getenv("GOOGLE_API_KEY"), base_url="https://api.openai.com/v1")
```

```python
MODEL = "gpt-4.1-nano"
db = "vector.db"
```

```python
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vectorstore = Chroma(persist_directory=db, embedding_function=embeddings)
```

```python
retriever = vectorstore.as_retriever()
llm = ChatGoogleGenerativeAI(model=MODEL_1,temperature=0)
```

```python
SYSTEM_PROMPT_TEMP = """
You are a knowledgeable, friendly assistant representing the company Insurellm.
You are chatting with a user about Insurellm.
If relevant, use the given context to answer any question.
If you don't know the answer, say so.
Context:
{context}
"""
```

```python
def answer(question,history):
    docs=retriever.invoke(question)
    context = "\n\n".join(doc.page_content for doc in docs)
    System_prompt=SYSTEM_PROMPT_TEMP.format(context=context)
    response=llm.invoke([SystemMessage(content=System_prompt),HumanMessage(content=question)])
    return response.content
```

```python
gr.ChatInterface(
    fn=answer,
    title="Receptionist at InsureLLM"
).launch(inbrowser=True)
```
